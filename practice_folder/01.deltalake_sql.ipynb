```python
%sql
-- SHOW CATALOGS;

USE CATALOG workspace;

CREATE TABLE product_info(
    product_id INT,
    product_name STRING,
    product_category STRING,
    product_price DOUBLE,
    product_quantity INT
)
USING DELTA;

```

```python
%sql
insert into product_info values (1, 'product1', 'category1', 10.0, 100),
                                (2, 'product2', 'category2', 20.0, 200),
                                (3, 'product3', 'category3', 30.0, 300),
                                (4, 'product4', 'category4', 40.0, 400),
                                (5, 'product5', 'category5', 50.0, 500);
```

```python
%sql
select * from product_info;
```
