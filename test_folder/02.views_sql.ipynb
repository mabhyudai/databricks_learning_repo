```python
%sql
CREATE TABLE product_availability(
    product_id INT,
    availability BOOLEAN,
    product_location STRING
)
```

```python
%sql
insert into product_availability values (1, true, 'warehouse1'), 
                                        (2, false, 'warehouse2'),
                                        (3, true, 'warehouse3'),
                                        (4, true, 'warehouse4'),
                                        (5, false, 'warehouse5');
```

```python
%sql
SELECT DISTINCT (pi.product_id), pi.product_name,pa.availability,pa.product_location FROM product_info pi
left join product_availability pa
ON pi.product_id = pa.product_id
ORDER BY pi.product_id;
```
